```python
# total sales
total_sales = df['Sales'].sum()
print('total_sales = ' , total_sales)
```

```python
# helps to find out the best selling category 
best_selling_category = df.groupby('Category')['Sales'].sum()
print("best_selling_category " , best_selling_category )
```

```python
# it help to show the which category help to make the more account transaction .
df.groupby('Category')['Sales'].count()
```

```python
#  SHOW THE BEST SALES BY REGION 
df.groupby('Region')['Sales'].sum()
```

```python
# showing sales by region with the help of bar graph .
import matplotlib.pyplot as plt

df.groupby('Region')['Sales'].sum().plot(kind='bar')
plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.show()
```

```python
# showing sales by category 

df.groupby('Category')['Sales'].sum().plot(kind='bar')
plt.title('sales by category')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.show()
```

```python
df['Segment'].value_counts()
```

```python
# showing sales by segment 

df.groupby('Segment')['Sales'].sum().plot(kind='bar')
plt.title('sales by segment')
plt.xlabel('Segment')
plt.ylabel('Sales')
plt.show()
```
